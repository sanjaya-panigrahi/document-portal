```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
import os
```

```python
file_path = os.path.join(os.getcwd(), "data", "Croma_Operations_Manual_v2.pdf")
```

```python
document = PyPDFLoader(file_path).load()
```

```python
len(document)
```

```python
text_splitter =  RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=150, length_function=len)
```

```python
docs = text_splitter.split_documents(document)
```

```python
 len(docs)
 len(document)
```

```python
docs
```

```python
docs[1].metadata
```

```python
docs[4].metadata
```

```python
docs[0].page_content
```

```python
from langchain_community.vectorstores import FAISS
```

```python
embedding_models.embed_documents([docs[1].page_content])
```

```python
vectorStore = FAISS.from_documents(docs, embedding_models)
```

```python
relevant_doc = vectorStore.similarity_search("What is Laptop Return Policy", k=5)
```

```python
relevant_doc[0].page_content
```

```python
rerank_doc = vectorStore.similarity_search_with_relevance_scores("What is Laptop Return Policy", k=5)
```

```python
rerank_doc
```

```python
retriever = vectorStore.as_retriever()
```

```python
retriever.invoke("What is Laptop Return Policy?")
```

```python
prompt_template = """
    Answer the question as truthfully as possible using the provided context. 
    If the answer is not containsufficient information, respond with :
    "i don't have enough information about it to answer that question."
    
    Context: {context}
    Question: {question}
    Answer: """
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt = PromptTemplate(
    template=prompt_template, input_variables=["context", "question"]
)
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
def format_docs(docs):
    return "\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.0-flash", temperature=0)
```

```python

rag_chain = (
        {"context": retriever | format_docs, "question": RunnablePassthrough()}
        | prompt
        | llm
        | parser
)
```

```python
rag_chain.invoke("What is Laptop Return Policy?")
```
